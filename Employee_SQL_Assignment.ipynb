This notebook demonstrates SQLite queries using a different employee dataset.

## 1. Create an employee table containing ID, name, department, salary, and city.

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE employees (
    employee_id INTEGER PRIMARY KEY,
    employee_name TEXT NOT NULL,
    department TEXT NOT NULL,
    salary INTEGER NOT NULL,
    city TEXT NOT NULL
)
""")

print("Employee table created successfully!")
```

## 2. Insert five employee records.

```python
employees = [
    (101, "Kavin", "Engineering", 62000, "Coimbatore"),
    (102, "Divya", "Finance", 48000, "Chennai"),
    (103, "Arun", "Engineering", 82000, "Bengaluru"),
    (104, "Meera", "Marketing", 57000, "Madurai"),
    (105, "Nithya", "Finance", 68000, "Coimbatore")
]

cursor.executemany(
    "INSERT INTO employees VALUES (?, ?, ?, ?, ?)",
    employees
)
conn.commit()
print("Five employee records inserted successfully!")
```

## 3. Display all employee records.

```python
cursor.execute("SELECT * FROM employees ORDER BY employee_id")
for row in cursor.fetchall():
    print(row)
```

## 4. Find employees whose salary is greater than 60,000.

```python
cursor.execute("""
SELECT employee_name, salary
FROM employees
WHERE salary > 60000
ORDER BY salary DESC
""")
print(cursor.fetchall())
```

## 5. Display all unique departments.

```python
cursor.execute("SELECT DISTINCT department FROM employees ORDER BY department")
print(cursor.fetchall())
```

## 6. Display the three employees with the highest salaries.

```python
cursor.execute("""
SELECT employee_name, salary
FROM employees
ORDER BY salary DESC
LIMIT 3
""")
print(cursor.fetchall())
```

## 7. Find Engineering employees earning more than 60,000, or employees in Finance.

```python
cursor.execute("""
SELECT employee_name, department, salary
FROM employees
WHERE (department = 'Engineering' AND salary > 60000)
   OR department = 'Finance'
ORDER BY employee_id
""")
print(cursor.fetchall())
```

## 8. Find employees whose names start with N.

```python
cursor.execute("""
SELECT employee_name
FROM employees
WHERE employee_name LIKE 'N%'
ORDER BY employee_name
""")
print(cursor.fetchall())
```

## 9. Calculate total, average, highest, and lowest salary.

```python
cursor.execute("""
SELECT SUM(salary), ROUND(AVG(salary), 2), MAX(salary), MIN(salary)
FROM employees
""")
print(cursor.fetchone())
```

## 10. Calculate the average salary of each department.

```python
cursor.execute("""
SELECT department, ROUND(AVG(salary), 2)
FROM employees
GROUP BY department
ORDER BY department
""")
print(cursor.fetchall())
```

## 11. Display departments whose average salary exceeds 55,000.

```python
cursor.execute("""
SELECT department, ROUND(AVG(salary), 2)
FROM employees
GROUP BY department
HAVING AVG(salary) > 55000
ORDER BY department
""")
print(cursor.fetchall())
```

## 12. Update Divya's salary to 52,000.

```python
cursor.execute("""
UPDATE employees
SET salary = 52000
WHERE employee_name = 'Divya'
""")
conn.commit()
print("Rows updated:", cursor.rowcount)
```

## 13. Delete the employee whose ID is 104.

```python
cursor.execute("DELETE FROM employees WHERE employee_id = 104")
conn.commit()
print("Rows deleted:", cursor.rowcount)
```

## 14. Create a department table and display employees with their department managers.

```python
cursor.execute("""
CREATE TABLE departments (
    department TEXT PRIMARY KEY,
    manager_name TEXT NOT NULL
)
""")

cursor.executemany(
    "INSERT INTO departments VALUES (?, ?)",
    [
        ("Engineering", "Suresh"),
        ("Finance", "Lakshmi"),
        ("Marketing", "Vijay")
    ]
)
conn.commit()

cursor.execute("""
SELECT e.employee_name, e.department, d.manager_name
FROM employees AS e
INNER JOIN departments AS d
ON e.department = d.department
ORDER BY e.employee_id
""")
print(cursor.fetchall())
```

## 15. Display all employees and managers, including employees without a matching department.

```python
cursor.execute("""
SELECT e.employee_name, e.department, d.manager_name
FROM employees AS e
LEFT JOIN departments AS d
ON e.department = d.department
ORDER BY e.employee_id
""")
print(cursor.fetchall())
```

## 16. Find employees whose salary is greater than the current average salary.

```python
cursor.execute("""
SELECT employee_name, salary
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees)
ORDER BY salary DESC
""")
print(cursor.fetchall())
```

## 17. Classify salaries of 65,000 or more as High Salary and others as Regular Salary.

```python
cursor.execute("""
SELECT employee_name, salary,
       CASE
           WHEN salary >= 65000 THEN 'High Salary'
           ELSE 'Regular Salary'
       END AS salary_category
FROM employees
ORDER BY employee_id
""")
print(cursor.fetchall())
```

## 18. Combine Engineering and Finance employee names without duplicates.

```python
cursor.execute("""
SELECT employee_name FROM employees WHERE department = 'Engineering'
UNION
SELECT employee_name FROM employees WHERE department = 'Finance'
ORDER BY employee_name
""")
print(cursor.fetchall())
```

## 19. Use a CTE to find employees earning more than 55,000.

```python
cursor.execute("""
WITH eligible_employees AS (
    SELECT employee_name, department, salary
    FROM employees
    WHERE salary > 55000
)
SELECT * FROM eligible_employees
ORDER BY salary DESC
""")
print(cursor.fetchall())
```

## 20. Rank employees according to salary from highest to lowest.

```python
cursor.execute("""
SELECT employee_name, salary,
       RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM employees
ORDER BY salary DESC
""")
print(cursor.fetchall())
```
