## Question 1: Normalize an AI-generated response

**Prompt:** Write a Python function to normalize a response by converting it to lowercase and removing extra spaces.

**Concept:** String methods and functions

````python
def normalize_text(text):
    return " ".join(text.lower().split())

response = "  Python  is Useful for Data Analysis!  "
normalized_response = normalize_text(response)

print("Original response:", response)
print("Normalized response:", normalized_response)
````

**Expected Output**
```text
Original response:   Python  is Useful for Data Analysis!  
Normalized response: python is useful for data analysis!
```

## Question 2: Compare a response with reference information

**Prompt:** Write a Python program to check whether the key terms in a response overlap with a reference statement.

**Concept:** Sets and string processing

````python
reference = "Python is commonly used for data analysis"
response = "Python is useful for data analysis"

reference_terms = set(normalize_text(reference).split())
response_terms = set(normalize_text(response).split())

matched_terms = reference_terms.intersection(response_terms)

print("Matched terms:", sorted(matched_terms))
print("Number of matched terms:", len(matched_terms))
````

**Expected Output**
```text
Matched terms: ['analysis', 'data', 'for', 'is', 'python']
Number of matched terms: 5
```

## Question 3: Calculate reference-term coverage

**Prompt:** Calculate the percentage of reference terms that also appear in the response.

**Concept:** Sets, arithmetic, and functions

**Note:** Term coverage measures word overlap only; it does not prove factual correctness.

````python
def calculate_coverage(reference, response):
    reference_terms = set(normalize_text(reference).split())
    response_terms = set(normalize_text(response).split())

    if not reference_terms:
        return 0.0

    matched = reference_terms.intersection(response_terms)
    return round((len(matched) / len(reference_terms)) * 100, 2)

reference = "Mars is known as the Red Planet"
response = "Mars is known as the Red Planet"

coverage = calculate_coverage(reference, response)
print("Reference-term coverage:", str(coverage) + "%")
````

**Expected Output**
```text
Reference-term coverage: 100.0%
```

## Question 4: Flag a possible unsupported claim

**Prompt:** Create a simple rule-based check that flags a response when it contains a numeric claim that conflicts with the reference example.

**Concept:** Conditional statements and string matching

**Note:** This is a small demonstration using a known example, not a general fact-checking solution.

````python
reference = "At sea level, water boils at about 100 degrees Celsius."
response = "At sea level, water boils at 150 degrees Celsius."

if "100 degrees" in reference.lower() and "150 degrees" in response.lower():
    print("Warning: The response may conflict with the reference.")
else:
    print("No conflict detected by this simple rule.")
````

**Expected Output**
```text
Warning: The response may conflict with the reference.
```

## Question 5: Assign a validation status

**Prompt:** Assign a basic status based on reference-term coverage.

**Concept:** Functions and if-else statements

**Rule used in this example:**
- 70% or more: `High term overlap`
- Below 70%: `Low term overlap`

This status measures overlap, not truth.

````python
def validation_status(coverage):
    if coverage >= 70:
        return "High term overlap"
    return "Low term overlap"

reference = "Paris is the capital city of France"
response = "Paris is the capital of France"

coverage = calculate_coverage(reference, response)
status = validation_status(coverage)

print("Coverage:", str(coverage) + "%")
print("Validation status:", status)
````

**Expected Output**
```text
Coverage: 83.33%
Validation status: High term overlap
```

## Question 6: Validate multiple AI responses

**Prompt:** Apply the validation process to several sample responses and show the results in a table.

**Concept:** Lists, loops, functions, and Pandas

````python
import pandas as pd

samples = [
    {
        "Question": "Which planet is called the Red Planet?",
        "Reference": "Mars is known as the Red Planet",
        "Response": "Mars is known as the Red Planet"
    },
    {
        "Question": "What is the capital of France?",
        "Reference": "Paris is the capital of France",
        "Response": "Paris is the capital of France"
    },
    {
        "Question": "At what temperature does water boil at sea level?",
        "Reference": "Water boils at about 100 degrees Celsius at sea level",
        "Response": "Water boils at 150 degrees Celsius at sea level"
    },
    {
        "Question": "What is Python used for?",
        "Reference": "Python is commonly used for data analysis",
        "Response": "Python is a programming language"
    }
]

results = []

for sample in samples:
    coverage = calculate_coverage(sample["Reference"], sample["Response"])
    results.append({
        "Question": sample["Question"],
        "Response": sample["Response"],
        "Term Coverage (%)": coverage,
        "Status": validation_status(coverage)
    })

results_df = pd.DataFrame(results)
results_df
````

## Question 7: Visualize validation results

**Prompt:** Create a bar chart showing term coverage for each sample response.

**Concept:** Data visualization using Matplotlib

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.bar(range(len(results_df)), results_df["Term Coverage (%)"])
plt.xticks(
    range(len(results_df)),
    ["Sample 1", "Sample 2", "Sample 3", "Sample 4"]
)
plt.title("Reference-Term Coverage of Sample AI Responses")
plt.xlabel("AI Response")
plt.ylabel("Term Coverage (%)")
plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()
````

## Conclusion

This notebook demonstrates a basic Python workflow for AI response validation assistance. It normalizes text, compares reference terms, calculates term coverage, flags one example conflict, summarizes multiple responses, and visualizes the results.

### Limitations
- Word overlap cannot confirm whether a statement is true.
- Correct paraphrases may be flagged because they use different words.
- Incorrect responses may still have high overlap with the reference.
- A real system should use semantic models, trusted evidence sources, labelled evaluation data, and human review where appropriate.
